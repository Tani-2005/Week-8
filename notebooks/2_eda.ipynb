```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("Libraries loaded")
```

```python
sales_df = pd.read_csv("cleaned_sales_data.csv")
churn_df = pd.read_csv("cleaned_customer_churn.csv")
house_df = pd.read_csv("cleaned_house_prices.csv")

print("Cleaned datasets loaded")
```

```python
print("Sales Data Shape:", sales_df.shape)
print("Churn Data Shape:", churn_df.shape)
print("House Data Shape:", house_df.shape)
```

```python
sales_df.hist(figsize=(12,8))
plt.suptitle("Sales Dataset Distributions")
plt.show()
```

```python
house_df.hist(figsize=(12,8))
plt.suptitle("House Prices Distributions")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(sales_df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
plt.title("Sales Data Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(house_df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
plt.title("House Prices Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Churn", data=churn_df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=sales_df.select_dtypes(include=['int64','float64']))
plt.xticks(rotation=90)
plt.title("Sales Data Outlier Analysis")
plt.show()
```

```python
if {'Price','Quantity'}.issubset(sales_df.columns):
    plt.figure(figsize=(6,4))
    sns.scatterplot(x="Price", y="Quantity", data=sales_df)
    plt.title("Price vs Quantity Relationship")
    plt.show()
```

```python
# Example if dataset has Date column

if "Date" in sales_df.columns:
    sales_df["Date"] = pd.to_datetime(sales_df["Date"])
    monthly_sales = sales_df.groupby(sales_df["Date"].dt.month)["Total_Sales"].sum()

    monthly_sales.plot(kind="line", figsize=(8,5))
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.show()
```
