```python
import pandas as pd
import numpy as np
import os

# Display settings
pd.set_option("display.max_columns", None)

print("Libraries loaded successfully")
```

```python
sales_df = pd.read_csv("sales_data.csv")
churn_df = pd.read_csv("customer_churn.csv")
house_df = pd.read_csv("house_prices.csv")

print("Datasets loaded successfully")
```

```python
print("SALES DATA")
display(sales_df.head())
print(sales_df.info())
```

```python
print("CUSTOMER CHURN DATA")
display(churn_df.head())
print(churn_df.info())
```

```python
print("HOUSE PRICES DATA")
display(house_df.head())
print(house_df.info())
```

```python
print("Sales Missing Values")
print(sales_df.isnull().sum())

print("\nChurn Missing Values")
print(churn_df.isnull().sum())

print("\nHouse Prices Missing Values")
print(house_df.isnull().sum())
```

```python
def clean_numeric(df):
    num_cols = df.select_dtypes(include=['int64','float64']).columns
    df[num_cols] = df[num_cols].fillna(df[num_cols].median())
    return df

def clean_categorical(df):
    cat_cols = df.select_dtypes(include=['object']).columns
    for col in cat_cols:
        df[col].fillna(df[col].mode()[0], inplace=True)
    return df
```

```python
sales_df = clean_numeric(sales_df)
sales_df = clean_categorical(sales_df)

churn_df = clean_numeric(churn_df)
churn_df = clean_categorical(churn_df)

house_df = clean_numeric(house_df)
house_df = clean_categorical(house_df)

print("Missing values handled")
```

```python
sales_df.drop_duplicates(inplace=True)
churn_df.drop_duplicates(inplace=True)
house_df.drop_duplicates(inplace=True)

print("Duplicates removed")
```

```python

if {'Quantity', 'Price'}.issubset(sales_df.columns):
    sales_df["Total_Sales"] = sales_df["Quantity"] * sales_df["Price"]

print("Feature engineering completed")
```

```python
print("Sales Shape:", sales_df.shape)
print("Churn Shape:", churn_df.shape)
print("House Shape:", house_df.shape)

print("Cleaning validation done")
```

```python
sales_df.to_csv("cleaned_sales_data.csv", index=False)
churn_df.to_csv("cleaned_customer_churn.csv", index=False)
house_df.to_csv("cleaned_house_prices.csv", index=False)

print("Cleaned datasets saved successfully")
```
