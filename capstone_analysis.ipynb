```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("Libraries loaded")
```

# Business Problem Statement

The objective of this project is to perform a complete end-to-end
business analysis using real-world datasets to generate actionable
insights and strategic recommendations.

The analysis focuses on:

- Sales performance optimization
- Customer behavior and churn understanding
- Price and market trend analysis

Goal: Convert data into business decisions.

```python
sales_df = pd.read_csv("cleaned_sales_data.csv")
churn_df = pd.read_csv("cleaned_customer_churn.csv")
house_df = pd.read_csv("cleaned_house_prices.csv")
```

```python
print("Sales Data:", sales_df.shape)
print("Customer Churn Data:", churn_df.shape)
print("House Prices Data:", house_df.shape)
```

```python
if "Date" in sales_df.columns:
    sales_df["Date"] = pd.to_datetime(sales_df["Date"])
    
    monthly_sales = sales_df.groupby(
        sales_df["Date"].dt.month)["Total_Sales"].sum()

    monthly_sales.plot(figsize=(8,5))
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(
    sales_df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Sales Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Churn", data=churn_df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(house_df["Price"], kde=True)
plt.title("House Price Distribution")
plt.show()
```

```python
if {'Price','Total_Sales'}.issubset(sales_df.columns):
    plt.figure(figsize=(6,4))
    sns.scatterplot(
        x="Price",
        y="Total_Sales",
        data=sales_df
    )
    plt.title("Price vs Total Sales")
    plt.show()
```

## Key Business Insights

- Sales performance shows clear seasonal trends.
- Strong correlations exist between pricing and revenue.
- Customer churn indicates potential retention challenges.
- Price distribution suggests market segmentation opportunities.
- Pricing strategy significantly impacts sales performance.
